### 4.1 선형회귀의 필요성

```python
import pandas as pd
import numpy as np

data = pd.read_csv("rec_ab_test.csv")
data.head()
```

```python
import statsmodels.formula.api as smf
result = smf.ols('watch_time~C(recommender)', data=data).fit() # watch_time = beta0 + beta1*challenger + epsilon
result.summary().tables[1]
```

```python
# patsy ver.
# 회귀 식에 활용되는 X와 design matrix의 구조를 볼 수 있음

import patsy
import statsmodels.api as sm

y, X = patsy.dmatrices(
    'watch_time ~ C(recommender)', #  'watch_time ~ C(recommender, Treatment(reference="control"))' ; control 설정 가능
    data=data,
    return_type='dataframe'
)

result = sm.OLS(y, X).fit()

result.summary().tables[1]
```

```python
(data.groupby("recommender")["watch_time"].mean()) # beta1 = mean(challenger) - mean(benchmark)
```

### 4.2 회귀분석 이론

```python
X_cols = ["credit_limit", "wage", "credit_score1", "credit_score2"]
X = risk_data[X_cols].assign(intercep=1) # intercep이라는 새로운 컬럼 붙이기, 절편으로 활용
y = risk_data["default"]

def regress(y, X):
    return np.linalg.inv(X.T.dot(X)).dot(X.T.dot(y)) # beta_hat​=(X⊤X)^(−1)X⊤y
beta = regress(y, X)
beta
```

```python
formula = 'default ~ credit_limit + wage + credit_score1 + credit_score2'
model = smf.ols(formula, data = risk_data).fit()
model.summary().tables[1]
```

```python
# T ~ X

debiasing_model = smf.ols( 
    'credit_limit ~ wage + credit_score1 + credit_score2',
    data = risk_data
).fit()

risk_data_deb = risk_data.assign( # 평균을 잔차에 더하기
    credit_limit_res=(debiasing_model.resid + risk_data["credit_limit"].mean())
)
```

```python
# Y ~ T_tilde

model_w_deb_data = smf.ols('default ~ credit_limit_res',
                           data = risk_data_deb).fit()

model_w_deb_data.summary().tables[1]
```

```python
# Y ~ X

denoising_model = smf.ols(
    'default ~ wage + credit_score1 + credit_score2',
    data = risk_data_deb
).fit()

risk_data_denoise = risk_data_deb.assign(
    default_res = denoising_model.resid + risk_data_deb["default"].mean()
)
```

```python
model_se = smf.ols(
    'default ~ wage + credit_score1 + credit_score2',
    data = risk_data
).fit()

print("SE regression:", model_se.bse["wage"]) 

model_wage_aux = smf.ols(
    'wage ~ credit_score1 + credit_score2',
    data = risk_data
).fit()

# model df: N-4
se_formula = (np.std(model_se.resid)
              /(np.std(model_wage_aux.resid)*np.sqrt(len(risk_data)-4)))

print("SE formula:  ", se_formula) # 얘가 작아지면 다중공선성이 생겨 본회귀의 SE가 커진다.
```

```python
model_w_orthogonal = smf.ols('default_res ~ credit_limit_res',
                             data = risk_data_denoise).fit()
model_w_orthogonal.summary().tables[1]
```

### 4.6 선형회귀에서의 비선형성

```python
spend_data = pd.read_csv('spend_data.csv')
spend_data.head()
```

```python
model_spend = smf.ols(
    'spend ~ np.sqrt(credit_limit)', data = spend_data
).fit()

model_spend.summary().tables[1]
```

```python
model_spend = smf.ols(
    'spend ~ np.sqrt(credit_limit) + wage', data = spend_data
).fit()

model_spend.summary().tables[1]
```

```python
debias_spend_model = smf.ols(f'np.sqrt(credit_limit)~wage',data=spend_data).fit()
denoise_spend_model = smf.ols(f'spend ~ wage', data = spend_data).fit()

credit_limit_sqrt_deb = (debias_spend_model.resid + np.sqrt(spend_data['credit_limit']).mean())
spend_den = denoise_spend_model.resid + spend_data['spend'].mean()

spend_data_deb = (spend_data
                  .assign(credit_limit_sqrt_deb = credit_limit_sqrt_deb, spend_den = spend_den))

final_model = smf.ols(f'spend_den ~ credit_limit_sqrt_deb', data=spend_data_deb).fit()
final_model.summary().tables[1]
```

```python
risk_data_dummies = (
    risk_data_rnd.join(pd.get_dummies(risk_data_rnd["credit_score1_buckets"],
                                      prefix="sb",
                                      drop_first=True))
)
risk_data_dummies.head()
```

```python
model = smf.ols("default ~ credit_limit + C(credit_score1_buckets)", data = risk_data_rnd).fit()
model.summary().tables[1]
```

```python
def regress(df, t, y):
    return smf.ols(f"{y}~{t}", data=df).fit().params[t]

effect_by_group = (risk_data_rnd.groupby("credit_score1_buckets").apply(regress, y="default", t="credit_limit"))

effect_by_group
```

```python
group_size = risk_data_rnd.groupby("credit_score1_buckets").size()
ate = (effect_by_group * group_size).sum() / group_size.sum()
float(ate)
```

```python
model = smf.ols("default ~ credit_limit * C(credit_score1_buckets)",
                data = risk_data_rnd).fit()
model.summary().tables[1]
```

```python
(model.params[model.params.index.str.contains("credit_limit:")]
 + model.params["credit_limit"]).round(9)
```

```python
np.random.seed(123)

# std(t) = 1
t1 = np.random.normal(0, 1, size=10^4)
df1 = pd.DataFrame(dict(
    t=t1,
    y=1*t1, # ATE = 1
    g=1
))

# std(t) = 2
t2 = np.random.normal(0, 2, size=10^4)
df2 = pd.DataFrame(dict(
    t=t2,
    y=2*t2, # ATE = 2
    g=2
))

df = pd.concat([df1, df2])
df.head()
```

```python
effect_by_group = df.groupby("g").apply(regress, y="y", t="t")
ate = (effect_by_group*df.groupby("g").size()).sum() / df.groupby("g").size().sum()
float(ate)
```

```python
model = smf.ols("y ~ t + C(g)", data=df).fit()
model.params
```
