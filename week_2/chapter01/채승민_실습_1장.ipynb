### Causal Quantities: An Example


```python
#oracle dataset

pd.DataFrame(dict(
    i= [1,2,3,4,5,6],
    y0=[200,120,300, 450,600,600],
    y1=[220,140,400, 500,600,800],
    t= [0,0,0,1,1,1],
    x= [0,0,1,0,0,1],
)).assign(
    y = lambda d: (d["t"]*d["y1"] + (1-d["t"])*d["y0"]).astype(int),
    te=lambda d: d["y1"] - d["y0"]
)
```

```python
#observable dataset

pd.DataFrame(dict(
    i= [1,2,3,4,5,6],
    y0=[200,120,300, np.nan, np.nan, np.nan,],
    y1=[np.nan, np.nan, np.nan, 500,600,800],
    t= [0,0,0,1,1,1],
    x= [0,0,1,0,0,1],
)).assign(
    y = lambda d: np.where(d["t"]==1, d["y1"], d["y0"]).astype(int),
    te=lambda d: d["y1"] - d["y0"]
)
```

### A Visual Guide to Bias


```python
plt.rc('font', size=20)
fig = plt.figure()

sns.lmplot(data=data,
           ci=None,
           x="avg_week_sales",
           y="weekly_amount_sold",
           scatter=False,
           height=4, aspect=2)

plt.scatter(x=data.query("is_on_sale==1")["avg_week_sales"],
            y=data.query("is_on_sale==1")["weekly_amount_sold"],
            label="on sale",
            color=color[0], alpha=.8, marker=marker[0])

plt.scatter(x=data.query("is_on_sale==0")["avg_week_sales"],
            y=data.query("is_on_sale==0")["weekly_amount_sold"],
            label="not on sale",
            color=color[2], alpha=.6, marker=marker[1])
plt.legend(fontsize="14")

```

서브집단별 회귀선 추가

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
slope1, intercept1 = np.polyfit(data.query("is_on_sale==1")["avg_week_sales"],
            data.query("is_on_sale==1")["weekly_amount_sold"], 1)
slope0, intercept0 = np.polyfit(data.query("is_on_sale==0")["avg_week_sales"],
            data.query("is_on_sale==0")["weekly_amount_sold"], 1)
# 첫 번째 그래프
axes[0].scatter(data.query("is_on_sale==1")["avg_week_sales"],
            data.query("is_on_sale==1")["weekly_amount_sold"],
            label="on sale",
            color=color[0], alpha=.8, marker=marker[0])
axes[0].plot(data.query("is_on_sale==1")["avg_week_sales"], slope1 * data.query("is_on_sale==1")["avg_week_sales"] + intercept1, color='red', label='Regression Line')

# 두 번째 그래프
axes[1].scatter(data.query("is_on_sale==0")["avg_week_sales"],
            data.query("is_on_sale==0")["weekly_amount_sold"],
            label="not on sale",
            color=color[2], alpha=.6, marker=marker[1])
axes[1].plot(data.query("is_on_sale==0")["avg_week_sales"], slope0 * data.query("is_on_sale==0")["avg_week_sales"] + intercept0, color='red', label='Regression Line')

plt.tight_layout()
plt.show()
```
